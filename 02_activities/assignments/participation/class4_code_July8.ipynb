```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL ##install as pillow if needed
import requests

# Generating random numbers for x and y variables
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
# plot both y variables on the same axis with labels and legend
fig, ax = plt.subplots(figsize=(5,3))
ax.plot (x, y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc = 'lower right', 
          frameon = True, 
          fontsize = 12, 
          ncol = 2,
          shadow = True)
```

```python
# Changing the location of the legend
fig, ax = plt.subplots(figsize= (5,3))
ax.plot(x, y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc = 'upper left', 
          bbox_to_anchor = (1,1))
fig.show()
```

```python
# Annotations with text function
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

# Modifying text annotations
ax.text(10,95,"This value is important!", 
        ha = 'center', 
        color = 'red', 
        size = 20)
```

```python
# Positioning text annotations
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1,5, ". Data: (1,5)", transform = ax.transData)
ax.text(0.5, 0.1, ". Axes: (0.5, 0.1)", transform = ax.transAxes)
ax.text(0.2, 0.2, ". Figure: (0.2, 0.2)", transform = fig.transFigure)
```

```python
# Annotating with arrows 
fig, ax = plt.subplots (figsize = (5,3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter (x, y2, label = "Person 2")
ax.legend (loc="lower right")

ax.annotate('This is important!', xy = (10, 95), xytext = (20, 94), 
            arrowprops = dict(facecolor = 'black'))
```

```python
# Annotating with arrows 
fig, ax = plt.subplots (figsize = (5,3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter (x, y2, label = "Person 2")
ax.legend (loc="lower right")

ax.annotate('This is important!', xy = (10, 95), xytext = (20, 94), 
            arrowprops = dict(arrowstyle = "wedge", 
                              color = "hotpink"))
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# Introducing subplots
fig, (ax1, ax2) = plt.subplots(ncols = 2, 
                               nrows = 1, 
                               figsize = (7,3))
ax1.scatter(x1, y1)
ax2.bar(x2, y2)
```

```python
# Activity: Customizing our plots
fig, (ax1, ax2) = plt.subplots(ncols = 2, 
                               nrows = 1, 
                               figsize = (7,3))
font1 = {'family':'serif', 'size': 10}
font2 = {'family':'serif', 'size': 12}

ax1.scatter(x1, y1, 
            marker = 'o', 
            color = 'blue')
ax2.bar(x2, y2, color = 'green')

ax1.set_title('Scatterplot', fontdict = font2, loc = 'left')
ax1.set_ylabel('Measure', fontdict = font1)
ax1.set_xlabel('Time', fontdict = font1)
ax2.set_title('Pirate Strength', fontdict = font2)
ax2.set_ylabel('Strength', fontdict = font1)
ax2.set_xlabel('Names', fontdict = font1)
ax1.xaxis.set_major_locator(plt.MultipleLocator(10))
```

```python
# Subplots without a grid arrangement
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4))

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2) 
someaxes['ax3'].plot(x1,y1)                                  
```

```python
# Layouts
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4))

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2) 
someaxes['ax3'].plot(x1,y1)  

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
# Modifying Layouts
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4), 
                                    layout = "constrained")

someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2) 
someaxes['ax3'].plot(x1,y1)  

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```
